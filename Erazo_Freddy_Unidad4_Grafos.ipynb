## Módulo 1 — Clase GrafoEcoVuelo

```python
class GrafoEcoVuelo:
    def __init__(self):
        # Lista de adyacencia: { ciudad: { vecino: km } }
        self.vertices = {}

    # 1. Añadir Ciudad (Agregar Vértice)
    def anadir_ciudad(self, ciudad):
        if ciudad not in self.vertices:
            self.vertices[ciudad] = {}
            print(f"Ciudad '{ciudad}' agregada al sistema.")
        else:
            print(f"'{ciudad}' ya existe en el sistema.")

    # 2. Eliminar Ciudad (Eliminar Vértice + sus aristas)
    def eliminar_ciudad(self, ciudad):
        if ciudad in self.vertices:
            del self.vertices[ciudad]
            for c in self.vertices:
                if ciudad in self.vertices[c]:
                    del self.vertices[c][ciudad]
            print(f"Ciudad '{ciudad}' y sus rutas eliminadas.")
        else:
            print(f"'{ciudad}' no existe en el sistema.")

    # 3. Añadir Ruta (Agregar Arista Ponderada)
    def anadir_ruta(self, origen, destino, km):
        if origen not in self.vertices or destino not in self.vertices:
            print("Una o ambas ciudades no existen.")
            return
        self.vertices[origen][destino] = km
        self.vertices[destino][origen] = km
        print(f"Ruta {origen} <-> {destino} ({km} km) agregada.")

    # 4. Eliminar Ruta (Eliminar Arista)
    def eliminar_ruta(self, origen, destino):
        if origen in self.vertices and destino in self.vertices[origen]:
            del self.vertices[origen][destino]
            del self.vertices[destino][origen]
            print(f"Ruta {origen} <-> {destino} eliminada.")
        else:
            print("La ruta no existe.")

    # 5. Buscar Ruta (Verificar Arista)
    def buscar_ruta(self, origen, destino):
        if origen in self.vertices and destino in self.vertices[origen]:
            print(f"Si existe ruta directa entre {origen} y {destino}.")
        else:
            print(f"No existe ruta directa entre {origen} y {destino}.")

    # 6. Buscar Ciudad (Verificar Vértice)
    def buscar_ciudad(self, ciudad):
        if ciudad in self.vertices:
            print(f"'{ciudad}' esta registrada en el sistema.")
        else:
            print(f"'{ciudad}' NO esta registrada en el sistema.")

    # 7. Destinos Directos (Obtener Adyacentes)
    def destinos_directos(self, ciudad):
        if ciudad not in self.vertices:
            print("Ciudad no existe.")
            return
        print(f"Destinos directos desde {ciudad}: {list(self.vertices[ciudad].keys())}")

    # 8. Consultar Distancia (Obtener Peso)
    def consultar_distancia(self, origen, destino):
        if origen in self.vertices and destino in self.vertices[origen]:
            print(f"Distancia {origen} <-> {destino}: {self.vertices[origen][destino]} km")
        else:
            print("No existe ruta directa entre esas ciudades.")

    # 9. Calcular Conectividad (Grado del Vértice)
    def calcular_conectividad(self, ciudad):
        if ciudad not in self.vertices:
            print("Ciudad no existe.")
            return
        print(f"Conectividad de {ciudad}: {len(self.vertices[ciudad])} rutas directas")

    # 10a. BFS — Búsqueda en Anchura
    def bfs(self, inicio):
        if inicio not in self.vertices:
            print("Ciudad no existe.")
            return
        visitados = []
        cola = [inicio]
        vistos = {inicio}
        while cola:
            actual = cola.pop(0)
            visitados.append(actual)
            for vecino in self.vertices[actual]:
                if vecino not in vistos:
                    vistos.add(vecino)
                    cola.append(vecino)
        print(f"BFS desde {inicio}: {visitados}")

    # 10b. DFS — Búsqueda en Profundidad
    def dfs(self, inicio):
        if inicio not in self.vertices:
            print("Ciudad no existe.")
            return
        visitados = []
        self._dfs(inicio, set(), visitados)
        print(f"DFS desde {inicio}: {visitados}")

    def _dfs(self, nodo, vistos, visitados):
        vistos.add(nodo)
        visitados.append(nodo)
        for vecino in self.vertices[nodo]:
            if vecino not in vistos:
                self._dfs(vecino, vistos, visitados)

    # Mostrar red completa
    def mostrar_red(self):
        print("\n--- RED AEREA ECOVUELO ---")
        for ciudad, rutas in self.vertices.items():
            print(f"  {ciudad} -> {rutas}")
        print("-" * 30)

print("Clase GrafoEcoVuelo cargada correctamente.")
```

## Módulo 3 — Interfaz de Escritorio (Tkinter + gráfico en vivo)

Ventana de escritorio independiente (`tkinter`, incluido con Python) dividida en dos paneles:

- **Panel izquierdo** — controles en tres pestañas: 🏙 Ciudades, ✈ Rutas, 🔍 Recorridos, más el cuadro de "Resultado".
- **Panel derecho** — el **mapa de la red**, más grande y con leyenda, que se dibuja y actualiza en vivo:
  - Al **añadir/eliminar** una ciudad o ruta, el mapa se redibuja al instante.
  - Al **elegir Origen/Destino** en la pestaña Rutas, esas dos ciudades se resaltan (verde = origen, naranja = destino) y la arista entre ellas se marca si existe.
  - Al ejecutar **BFS/DFS**, el camino recorrido se resalta paso a paso sobre el mismo mapa.
- Todas las operaciones se hacen con **botones, campos de texto y listas desplegables** — no se usa `input()` de consola.
- Mientras la ventana está abierta la celda sigue "ejecutando"; al **cerrar la ventana**, la celda termina y el control vuelve al notebook.
- Usa el mismo objeto `grafo` (clase `GrafoEcoVuelo` del Módulo 1) y sus datos iniciales del Módulo 2.

> Ejecuta la celda de abajo para abrir la ventana.


```python
# ── Módulo 7: Interfaz de Escritorio (Tkinter) con mapa en vivo ──────────────
import tkinter as tk
from tkinter import ttk

import heapq
import numpy as np
from matplotlib.figure import Figure
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg
import matplotlib.patches as mpatches
import matplotlib.patheffects as pe

COLOR_OK      = "#1A8A60"
COLOR_ERR     = "#C94A0C"
COLOR_AVISO   = "#B07010"
COLOR_INFO    = "#1557C0"
COLOR_ORIGEN  = "#1A8A60"
COLOR_DESTINO = "#B07010"
COLOR_ARISTA  = "#1557C0"
COLOR_SELEC   = "#7A3FA0"

PLACEHOLDER = "— selecciona —"

# Posiciones geográficas base (Sierra Ecuador, coordenadas normalizadas 0-1)
_POS_BASE = {
    "Ibarra":    (0.18, 0.88),
    "Quito":     (0.42, 0.68),
    "Latacunga": (0.46, 0.46),
    "Ambato":    (0.70, 0.30),
    "Guaranda":  (0.22, 0.16),
    "Riobamba":  (0.52, 0.06),
}


class VentanaEcoVuelo:
    def __init__(self, grafo):
        self.grafo = grafo

        # Estado de resaltado del mapa (se actualiza según la interacción del usuario)
        self._h_origen = None
        self._h_destino = None
        self._h_camino = None
        self._h_color_camino = COLOR_ARISTA

        # Posiciones ya calculadas en el mapa (se conservan entre redibujos para
        # que las ciudades no "salten" de lugar cada vez que se actualiza el mapa)
        self._pos_cache = dict(_POS_BASE)

        self.root = tk.Tk()
        self.root.title("✈ EcoVuelo — Sistema de Rutas Aéreas (Escritorio)")
        self.root.geometry("1180x720")
        self.root.minsize(980, 600)

        estilo = ttk.Style(self.root)
        try:
            estilo.theme_use("clam")
        except tk.TclError:
            pass

        # ── layout general: panel izquierdo (controles) | panel derecho (mapa) ──
        contenedor = tk.Frame(self.root)
        contenedor.pack(fill="both", expand=True)

        col_izq = tk.Frame(contenedor, width=380)
        col_izq.pack(side="left", fill="y")
        col_izq.pack_propagate(False)

        col_der = tk.Frame(contenedor, bg="#FFFFFF")
        col_der.pack(side="left", fill="both", expand=True)

        # ── encabezado (panel izquierdo) ──────────────────────────────────
        header = tk.Frame(col_izq, pady=10, padx=14)
        header.pack(fill="x")
        tk.Label(header, text="✈ EcoVuelo", font=("Georgia", 16),
                  fg=COLOR_INFO).pack(anchor="w")
        tk.Label(header, text="Sistema de Rutas Aéreas · Sierra del Ecuador",
                  font=("Segoe UI", 9), fg="#999999").pack(anchor="w")

        # ── pestañas (panel izquierdo) ────────────────────────────────────
        self.tabs = ttk.Notebook(col_izq)
        self.tabs.pack(fill="x", padx=14, pady=(4, 6))

        self.tab_ciudades   = ttk.Frame(self.tabs, padding=14)
        self.tab_rutas      = ttk.Frame(self.tabs, padding=14)
        self.tab_recorridos = ttk.Frame(self.tabs, padding=14)
        self.tabs.add(self.tab_ciudades, text="🏙 Ciudades")
        self.tabs.add(self.tab_rutas, text="✈ Rutas")
        self.tabs.add(self.tab_recorridos, text="🔍 Recorridos")

        self._armar_tab_ciudades()
        self._armar_tab_rutas()
        self._armar_tab_recorridos()

        # ── panel de resultado (panel izquierdo) ──────────────────────────
        tk.Label(col_izq, text="RESULTADO", font=("Segoe UI", 8, "bold"),
                  fg="#aaaaaa").pack(anchor="w", padx=16, pady=(6, 0))
        self.txt_resultado = tk.Text(col_izq, height=9, wrap="word",
                                      font=("Segoe UI", 10), padx=8, pady=6,
                                      state="disabled", relief="solid", borderwidth=1)
        self.txt_resultado.pack(fill="both", expand=True, padx=14, pady=(0, 14))
        self.txt_resultado.tag_configure("ok", foreground=COLOR_OK)
        self.txt_resultado.tag_configure("err", foreground=COLOR_ERR)
        self.txt_resultado.tag_configure("aviso", foreground=COLOR_AVISO)
        self.txt_resultado.tag_configure("info", foreground=COLOR_INFO)

        # ── mapa del grafo (panel derecho, se dibuja y actualiza en vivo) ──
        marco_mapa = tk.Frame(col_der, bg="#FFFFFF")
        marco_mapa.pack(fill="both", expand=True, padx=6, pady=6)

        self.fig = Figure(figsize=(8, 7.5), dpi=100)
        self.fig.patch.set_facecolor("#FFFFFF")
        self.fig.subplots_adjust(left=0.02, right=0.98, top=0.98, bottom=0.02)
        self.ax = self.fig.add_subplot(111)

        self.canvas = FigureCanvasTkAgg(self.fig, master=marco_mapa)
        self.canvas.get_tk_widget().pack(fill="both", expand=True)

        self._dibujar()

    # ── helpers de mensajes ────────────────────────────────────────────────
    def _mostrar(self, tag, icono, mensaje):
        self.txt_resultado.configure(state="normal")
        self.txt_resultado.delete("1.0", "end")
        self.txt_resultado.insert("end", f"{icono}  {mensaje}", tag)
        self.txt_resultado.configure(state="disabled")

    def _ok(self, m):    self._mostrar("ok", "✔", m)
    def _err(self, m):   self._mostrar("err", "✘", m)
    def _aviso(self, m): self._mostrar("aviso", "⚠", m)
    def _info(self, m):  self._mostrar("info", "ℹ", m)

    def _opciones_ciudades(self):
        return [PLACEHOLDER] + sorted(self.grafo.vertices.keys())

    def _refrescar_dropdowns(self):
        opts = self._opciones_ciudades()
        for combo in (self.cb_origen, self.cb_destino, self.cb_bfs, self.cb_dfs):
            actual = combo.get()
            combo["values"] = opts
            combo.set(actual if actual in opts else PLACEHOLDER)

    def _posicion(self, ciudad):
        """Posición de una ciudad en el mapa. Las ciudades base tienen una
        ubicación fija; una ciudad nueva se coloca en el centroide de las
        ciudades con las que se conecta (para que aparezca "entre" sus
        conexiones y no en un punto cualquiera del lienzo). Una vez calculada,
        la posición se guarda para que no cambie en redibujos posteriores."""
        if ciudad in self._pos_cache:
            return self._pos_cache[ciudad]

        vecinos = self.grafo.vertices.get(ciudad, {})
        vecinos_ubicados = [self._pos_cache[v] for v in vecinos if v in self._pos_cache]

        rng = np.random.default_rng(abs(hash(ciudad)) % (2 ** 31))
        if vecinos_ubicados:
            cx = sum(p[0] for p in vecinos_ubicados) / len(vecinos_ubicados)
            cy = sum(p[1] for p in vecinos_ubicados) / len(vecinos_ubicados)
            # pequeño desplazamiento para no quedar exactamente encima de una
            # línea o de otra ciudad cuando hay varias conexiones alineadas
            angulo = rng.uniform(0, 2 * np.pi)
            radio_offset = 0.06 if len(vecinos_ubicados) > 1 else 0.10
            x = cx + radio_offset * np.cos(angulo)
            y = cy + radio_offset * np.sin(angulo)
        else:
            # sin conexiones todavía: se ubica en un punto libre pseudoaleatorio
            x = rng.uniform(0.15, 0.85)
            y = rng.uniform(0.15, 0.85)

        x = float(min(0.90, max(0.10, x)))
        y = float(min(0.90, max(0.10, y)))
        self._pos_cache[ciudad] = (x, y)
        return self._pos_cache[ciudad]

    def _lado_etiqueta(self, ciudad, pos):
        """Decide si el nombre de la ciudad va arriba o abajo del nodo,
        según dónde estén la mayoría de sus vecinos (para no cruzar las líneas)."""
        vecinos = self.grafo.vertices.get(ciudad, {})
        if not vecinos:
            return "arriba"
        y = pos[ciudad][1]
        arriba = sum(1 for v in vecinos if pos[v][1] > y)
        abajo = sum(1 for v in vecinos if pos[v][1] < y)
        return "abajo" if arriba > abajo else "arriba"

    # ── dibujo del mapa ──────────────────────────────────────────────────────
    def _dibujar(self):
        """Redibuja el grafo completo aplicando el resaltado actual
        (origen/destino elegidos en Rutas, o el camino de un BFS/DFS)."""
        ax = self.ax
        ax.clear()
        ax.set_facecolor("#FFFFFF")
        ax.set_xlim(-0.08, 1.08)
        ax.set_ylim(-0.10, 1.08)
        ax.axis("off")

        if not self.grafo.vertices:
            ax.text(0.5, 0.5, "La red está vacía", ha="center", va="center",
                    fontsize=15, color="#aaaaaa")
            self.canvas.draw()
            return

        pos = {c: self._posicion(c) for c in self.grafo.vertices}
        camino = self._h_camino or []
        pares_camino = set(frozenset(par) for par in zip(camino, camino[1:]))

        # ── aristas ───────────────────────────────────────────────────────
        vistas = set()
        for origen, rutas in self.grafo.vertices.items():
            for destino, km in rutas.items():
                par = frozenset((origen, destino))
                if par in vistas:
                    continue
                vistas.add(par)
                x1, y1 = pos[origen]
                x2, y2 = pos[destino]

                color_arista, alpha_arista, lw_arista = "#5B8DEF", 0.75, 1.6
                if par in pares_camino:
                    color_arista = self._h_color_camino
                    alpha_arista, lw_arista = 1.0, 3.6
                elif self._h_origen and self._h_destino and par == frozenset((self._h_origen, self._h_destino)):
                    color_arista, alpha_arista, lw_arista = COLOR_SELEC, 1.0, 3.2

                ax.plot([x1, x2], [y1, y2], color=color_arista, alpha=alpha_arista,
                        lw=lw_arista, solid_capstyle="round", zorder=1)

                mx, my = (x1 + x2) / 2, (y1 + y2) / 2
                ax.text(mx, my, f"{km} km", ha="center", va="center", fontsize=9.5,
                        color="#5A6B85", zorder=2,
                        bbox=dict(facecolor="white", edgecolor="none", alpha=0.9,
                                  boxstyle="round,pad=0.15"))

        # ── nodos ─────────────────────────────────────────────────────────
        radio = 0.052
        for ciudad, (x, y) in pos.items():
            color_nodo = "#1557C0"
            if ciudad in camino:
                orden = camino.index(ciudad)
                intensidad = max(0.35, 1.0 - 0.12 * orden)
                base = self._h_color_camino.lstrip("#")
                r, g, b = (int(base[i:i + 2], 16) for i in (0, 2, 4))
                color_nodo = f"#{int(r * intensidad):02x}{int(g * intensidad):02x}{int(b * intensidad):02x}"
            elif ciudad == self._h_origen:
                color_nodo = COLOR_ORIGEN
            elif ciudad == self._h_destino:
                color_nodo = COLOR_DESTINO

            sombra = mpatches.Circle((x + 0.007, y - 0.007), radio, color="#00000022", zorder=2)
            ax.add_patch(sombra)
            circulo = mpatches.Circle((x, y), radio, color=color_nodo, zorder=3)
            ax.add_patch(circulo)

            ax.text(x, y, ciudad[0], ha="center", va="center", fontsize=13,
                    fontweight="bold", color="white", zorder=4)

            lado = self._lado_etiqueta(ciudad, pos)
            ly = y + radio + 0.030 if lado == "arriba" else y - radio - 0.030
            va = "bottom" if lado == "arriba" else "top"
            ax.text(x, ly, ciudad, ha="center", va=va, fontsize=10.5,
                    color="#0E1C30", fontweight="600", zorder=4,
                    path_effects=[pe.withStroke(linewidth=3, foreground="#FFFFFF")])

        self.canvas.draw()

    def _limpiar_resaltado(self):
        self._h_origen = None
        self._h_destino = None
        self._h_camino = None

    def _ruta_mas_corta(self, origen, destino):
        """Dijkstra: camino de menor distancia total (en km) entre dos ciudades,
        aunque no exista un vuelo directo entre ellas. Devuelve (camino, km_total)
        o (None, None) si no hay ninguna conexión posible."""
        distancias = {origen: 0}
        previos = {}
        visitados = set()
        cola = [(0, origen)]
        while cola:
            d, actual = heapq.heappop(cola)
            if actual in visitados:
                continue
            visitados.add(actual)
            if actual == destino:
                break
            for vecino, km in self.grafo.vertices[actual].items():
                nueva_d = d + km
                if vecino not in distancias or nueva_d < distancias[vecino]:
                    distancias[vecino] = nueva_d
                    previos[vecino] = actual
                    heapq.heappush(cola, (nueva_d, vecino))

        if destino not in distancias:
            return None, None

        camino = [destino]
        while camino[-1] != origen:
            camino.append(previos[camino[-1]])
        camino.reverse()
        return camino, distancias[destino]

    # ── Tab 1: Ciudades ─────────────────────────────────────────────────────
    def _armar_tab_ciudades(self):
        t = self.tab_ciudades
        tk.Label(t, text="Nombre de la ciudad:").grid(row=0, column=0, sticky="w", pady=(0, 4))
        self.txt_ciudad = ttk.Entry(t, width=28)
        self.txt_ciudad.grid(row=1, column=0, columnspan=3, sticky="w", pady=(0, 10))

        ttk.Button(t, text="➕ Añadir", command=self._on_add_ciudad).grid(row=2, column=0, padx=(0, 6), pady=2)
        ttk.Button(t, text="🗑 Eliminar", command=self._on_del_ciudad).grid(row=2, column=1, padx=6, pady=2)
        ttk.Button(t, text="🔍 Buscar", command=self._on_find_ciudad).grid(row=2, column=2, padx=6, pady=2)
        ttk.Button(t, text="📊 Conectividad", command=self._on_conectividad).grid(row=3, column=0, padx=(0, 6), pady=2, sticky="w")
        ttk.Button(t, text="🗺 Destinos directos", command=self._on_destinos).grid(row=3, column=1, columnspan=2, padx=6, pady=2, sticky="w")

    def _nombre_ciudad(self):
        return self.txt_ciudad.get().strip().title()

    def _on_add_ciudad(self):
        n = self._nombre_ciudad()
        if not n:
            self._err("Escribe el nombre de la ciudad."); return
        if n in self.grafo.vertices:
            self._aviso(f"'{n}' ya existía en el sistema."); return

        conexiones = self._pedir_conexiones(n)
        if conexiones is None:
            self._info("Operación cancelada. La ciudad no fue agregada.")
            return

        self.grafo.anadir_ciudad(n)
        for destino, km in conexiones:
            self.grafo.anadir_ruta(n, destino, km)

        if conexiones:
            detalle = ", ".join(f"{c} ({km} km)" for c, km in conexiones)
            self._ok(f"Ciudad '{n}' agregada con {len(conexiones)} conexión(es): {detalle}.")
        else:
            self._ok(f"Ciudad '{n}' agregada como primera ciudad del sistema (sin conexiones aún).")

        self.txt_ciudad.delete(0, "end")
        self._limpiar_resaltado()
        if conexiones:
            self._h_camino = [n] + [c for c, _ in conexiones]
            self._h_color_camino = COLOR_OK
        else:
            self._h_origen = n
        self._refrescar_dropdowns()
        self._dibujar()

    def _pedir_conexiones(self, nueva_ciudad):
        """Ventana emergente: pregunta con qué ciudades existentes se conecta
        la nueva ciudad y a qué distancia (km). Devuelve una lista de
        (ciudad_vecina, km) o None si el usuario cancela."""
        resultado = {"conexiones": None}
        existentes = sorted(self.grafo.vertices.keys())
        conexiones_tmp = []

        dialog = tk.Toplevel(self.root)
        dialog.title(f"Ubicar '{nueva_ciudad}'")
        dialog.geometry("440x440")
        dialog.transient(self.root)
        dialog.grab_set()
        dialog.resizable(False, False)

        tk.Label(dialog, text=f"¿Dónde se ubica '{nueva_ciudad}'?",
                 font=("Segoe UI", 11, "bold")).pack(anchor="w", padx=16, pady=(16, 2))
        if self.grafo.vertices:
            texto_ayuda = ("Indica a qué ciudad(es) se conecta y a qué distancia (km).\n"
                            "Debes agregar al menos una conexión: una ciudad no puede\n"
                            "crearse aislada del resto de la red.")
        else:
            texto_ayuda = ("Esta será la primera ciudad del sistema, así que puede\n"
                            "crearse sin conexiones todavía.")
        tk.Label(
            dialog, text=texto_ayuda,
            font=("Segoe UI", 8), fg="#888888", justify="left",
        ).pack(anchor="w", padx=16, pady=(0, 10))

        lbl_advertencia = tk.Label(dialog, text="", font=("Segoe UI", 8, "bold"),
                                    fg=COLOR_ERR, justify="left")

        marco_form = tk.Frame(dialog)
        marco_form.pack(fill="x", padx=16)

        tk.Label(marco_form, text="Ciudad existente:").grid(row=0, column=0, sticky="w")
        tk.Label(marco_form, text="Distancia (km):").grid(row=0, column=1, sticky="w", padx=(10, 0))

        cb_ciudad = ttk.Combobox(marco_form, values=existentes, state="readonly", width=16)
        if existentes:
            cb_ciudad.set(existentes[0])
        cb_ciudad.grid(row=1, column=0, sticky="w", pady=(2, 8))

        spin_km_dlg = ttk.Spinbox(marco_form, from_=1, to=9999, width=8)
        spin_km_dlg.set(100)
        spin_km_dlg.grid(row=1, column=1, sticky="w", padx=(10, 0), pady=(2, 8))

        lista = tk.Listbox(dialog, height=9)
        lista.pack(fill="both", expand=True, padx=16, pady=(4, 4))

        def _agregar():
            if not existentes:
                return
            c = cb_ciudad.get()
            if not c:
                return
            try:
                km = int(spin_km_dlg.get())
            except ValueError:
                return
            if km <= 0:
                return
            for i, (cc, _) in enumerate(conexiones_tmp):
                if cc == c:
                    conexiones_tmp[i] = (c, km)
                    lista.delete(i)
                    lista.insert(i, f"{nueva_ciudad} ↔ {c}  ({km} km)")
                    return
            conexiones_tmp.append((c, km))
            lista.insert("end", f"{nueva_ciudad} ↔ {c}  ({km} km)")
            lbl_advertencia.config(text="")

        def _quitar():
            sel = lista.curselection()
            if not sel:
                return
            idx = sel[0]
            lista.delete(idx)
            conexiones_tmp.pop(idx)

        ttk.Button(marco_form, text="➕ Agregar conexión", command=_agregar).grid(
            row=1, column=2, sticky="w", padx=(10, 0))

        if not existentes:
            tk.Label(dialog, text="(Todavía no hay otras ciudades registradas.)",
                     font=("Segoe UI", 8), fg="#aaaaaa").pack(anchor="w", padx=16)
            cb_ciudad.configure(state="disabled")

        ttk.Button(dialog, text="🗑 Quitar seleccionada", command=_quitar).pack(
            anchor="w", padx=16, pady=(0, 8))

        lbl_advertencia.pack(anchor="w", padx=16, pady=(0, 4))

        marco_final = tk.Frame(dialog)
        marco_final.pack(fill="x", padx=16, pady=(4, 16))

        def _guardar():
            if self.grafo.vertices and not conexiones_tmp:
                lbl_advertencia.config(
                    text="⚠ Agrega al menos una conexión antes de guardar.\n"
                         "Una ciudad no puede crearse aislada del resto de la red."
                )
                return
            resultado["conexiones"] = list(conexiones_tmp)
            dialog.destroy()

        def _cancelar():
            resultado["conexiones"] = None
            dialog.destroy()

        ttk.Button(marco_final, text="💾 Guardar ciudad", command=_guardar).pack(side="left")
        ttk.Button(marco_final, text="✖ Cancelar", command=_cancelar).pack(side="left", padx=(10, 0))

        dialog.protocol("WM_DELETE_WINDOW", _cancelar)
        dialog.wait_visibility()
        dialog.focus_set()
        self.root.wait_window(dialog)
        return resultado["conexiones"]

    def _on_del_ciudad(self):
        n = self._nombre_ciudad()
        if not n:
            self._err("Escribe el nombre de la ciudad."); return
        if n not in self.grafo.vertices:
            self._err(f"'{n}' no está en el sistema."); return
        rutas = list(self.grafo.vertices[n].keys())
        self.grafo.eliminar_ciudad(n)
        self._pos_cache.pop(n, None)
        extra = f" (rutas eliminadas: {', '.join(rutas)})" if rutas else ""
        self._ok(f"Ciudad '{n}' eliminada.{extra}")
        self._limpiar_resaltado()
        self._refrescar_dropdowns()
        self._dibujar()

    def _on_find_ciudad(self):
        n = self._nombre_ciudad()
        if not n:
            self._err("Escribe el nombre de la ciudad."); return
        if n in self.grafo.vertices:
            self._ok(f"'{n}' está registrada en el sistema.")
            self._h_origen, self._h_destino, self._h_camino = n, None, None
            self._dibujar()
        else:
            self._aviso(f"'{n}' NO está registrada en el sistema.")

    def _on_conectividad(self):
        n = self._nombre_ciudad()
        if not n:
            self._err("Escribe el nombre de la ciudad."); return
        if n not in self.grafo.vertices:
            self._err(f"'{n}' no existe en el sistema."); return
        g = len(self.grafo.vertices[n])
        barra = "█" * g + "░" * max(0, 6 - g)
        self._info(f"Conectividad de '{n}': {g} ruta(s)  {barra}")
        self._h_origen, self._h_destino, self._h_camino = n, None, None
        self._dibujar()

    def _on_destinos(self):
        n = self._nombre_ciudad()
        if not n:
            self._err("Escribe el nombre de la ciudad."); return
        if n not in self.grafo.vertices:
            self._err(f"'{n}' no existe en el sistema."); return
        vecinos = self.grafo.vertices[n]
        if not vecinos:
            self._aviso(f"'{n}' no tiene destinos directos aún."); return
        texto = ", ".join(f"{v} ({km} km)" for v, km in sorted(vecinos.items()))
        self._info(f"Destinos directos desde '{n}': {texto}")
        self._h_origen, self._h_destino, self._h_camino = n, None, [n] + list(vecinos.keys())
        self._h_color_camino = COLOR_INFO
        self._dibujar()

    # ── Tab 2: Rutas ─────────────────────────────────────────────────────────
    def _armar_tab_rutas(self):
        t = self.tab_rutas
        tk.Label(t, text="Origen:").grid(row=0, column=0, sticky="w")
        tk.Label(t, text="Destino:").grid(row=0, column=1, sticky="w", padx=(14, 0))

        opts = self._opciones_ciudades()
        self.cb_origen = ttk.Combobox(t, values=opts, state="readonly", width=18)
        self.cb_origen.set(PLACEHOLDER)
        self.cb_origen.grid(row=1, column=0, sticky="w", pady=(0, 10))
        self.cb_origen.bind("<<ComboboxSelected>>", self._on_seleccion_od)

        self.cb_destino = ttk.Combobox(t, values=opts, state="readonly", width=18)
        self.cb_destino.set(PLACEHOLDER)
        self.cb_destino.grid(row=1, column=1, sticky="w", padx=(14, 0), pady=(0, 10))
        self.cb_destino.bind("<<ComboboxSelected>>", self._on_seleccion_od)

        tk.Label(t, text="Distancia (km):").grid(row=2, column=0, sticky="w")
        self.spin_km = ttk.Spinbox(t, from_=1, to=9999, width=10)
        self.spin_km.set(100)
        self.spin_km.grid(row=3, column=0, sticky="w", pady=(0, 10))

        ttk.Button(t, text="➕ Añadir / actualizar", command=self._on_add_ruta).grid(row=4, column=0, padx=(0, 6), pady=2, sticky="w")
        ttk.Button(t, text="🗑 Eliminar", command=self._on_del_ruta).grid(row=4, column=1, padx=6, pady=2, sticky="w")
        ttk.Button(t, text="🔍 Buscar ruta", command=self._on_find_ruta).grid(row=5, column=0, padx=(0, 6), pady=2, sticky="w")
        ttk.Button(t, text="📏 Distancia", command=self._on_distancia).grid(row=5, column=1, padx=6, pady=2, sticky="w")

    def _origen_destino(self):
        return self.cb_origen.get(), self.cb_destino.get()

    def _on_seleccion_od(self, event=None):
        """Se dispara cada vez que el usuario elige Origen o Destino: resalta
        ambas ciudades (y la ruta entre ellas si existe) en el mapa, en vivo."""
        o, d = self._origen_destino()
        self._h_camino = None
        self._h_origen = o if o != PLACEHOLDER else None
        self._h_destino = d if d != PLACEHOLDER else None
        self._dibujar()

    def _on_add_ruta(self):
        o, d = self._origen_destino()
        if PLACEHOLDER in (o, d):
            self._err("Selecciona origen y destino."); return
        if o == d:
            self._err("Origen y destino deben ser ciudades distintas."); return
        km = int(self.spin_km.get())
        existe = d in self.grafo.vertices.get(o, {})
        self.grafo.anadir_ruta(o, d, km)
        if existe:
            self._ok(f"Ruta {o} ↔ {d} actualizada a {km} km.")
        else:
            self._ok(f"Ruta {o} ↔ {d} ({km} km) agregada.")
        self._dibujar()

    def _on_del_ruta(self):
        o, d = self._origen_destino()
        if PLACEHOLDER in (o, d):
            self._err("Selecciona origen y destino."); return
        if d not in self.grafo.vertices.get(o, {}):
            self._err(f"No existe ruta directa {o} ↔ {d}."); return
        self.grafo.eliminar_ruta(o, d)
        self._ok(f"Ruta {o} ↔ {d} eliminada.")
        self._dibujar()

    def _on_find_ruta(self):
        o, d = self._origen_destino()
        if PLACEHOLDER in (o, d):
            self._err("Selecciona origen y destino."); return
        if o == d:
            self._err("Origen y destino deben ser ciudades distintas."); return

        if d in self.grafo.vertices.get(o, {}):
            self._ok(f"Sí existe ruta directa {o} ↔ {d} ({self.grafo.vertices[o][d]} km).")
            self._h_origen, self._h_destino, self._h_camino = o, d, None
            self._dibujar()
            return

        camino, total_km = self._ruta_mas_corta(o, d)
        if camino is None:
            self._err(f"No hay ruta directa entre {o} y {d}, ni tampoco una ruta indirecta (ciudades no conectadas).")
            self._h_origen, self._h_destino, self._h_camino = o, d, None
            self._dibujar()
            return

        tramos = " + ".join(
            f"{a}-{b} ({self.grafo.vertices[a][b]} km)" for a, b in zip(camino, camino[1:])
        )
        self._aviso(
            f"No hay ruta directa entre {o} y {d}. La ruta más adecuada (menor distancia) es: "
            f"{' → '.join(camino)}  =  {tramos}  →  total {total_km} km."
        )
        self._h_origen, self._h_destino = None, None
        self._h_camino, self._h_color_camino = camino, COLOR_SELEC
        self._dibujar()

    def _on_distancia(self):
        o, d = self._origen_destino()
        if PLACEHOLDER in (o, d):
            self._err("Selecciona origen y destino."); return
        if o == d:
            self._err("Origen y destino deben ser ciudades distintas."); return

        if d in self.grafo.vertices.get(o, {}):
            self._ok(f"Distancia {o} ↔ {d}: {self.grafo.vertices[o][d]} km (vuelo directo).")
            self._h_origen, self._h_destino, self._h_camino = o, d, None
            self._dibujar()
            return

        camino, total_km = self._ruta_mas_corta(o, d)
        if camino is None:
            self._err(f"No existe ninguna ruta entre {o} y {d} (ciudades no conectadas).")
            self._h_origen, self._h_destino, self._h_camino = o, d, None
            self._dibujar()
            return

        tramos = " + ".join(
            f"{a}-{b} ({self.grafo.vertices[a][b]} km)" for a, b in zip(camino, camino[1:])
        )
        self._aviso(
            f"No hay vuelo directo {o} ↔ {d}. Distancia más corta vía {' → '.join(camino)}: "
            f"{tramos}  →  total {total_km} km ({len(camino) - 1} escala(s))."
        )
        self._h_origen, self._h_destino = None, None
        self._h_camino, self._h_color_camino = camino, COLOR_SELEC
        self._dibujar()

    # ── Tab 3: Recorridos ────────────────────────────────────────────────────
    def _armar_tab_recorridos(self):
        t = self.tab_recorridos
        opts = self._opciones_ciudades()

        tk.Label(t, text="BFS — Búsqueda en Anchura", font=("Segoe UI", 10, "bold")).grid(row=0, column=0, sticky="w")
        self.cb_bfs = ttk.Combobox(t, values=opts, state="readonly", width=18)
        self.cb_bfs.set(PLACEHOLDER)
        self.cb_bfs.grid(row=1, column=0, sticky="w", pady=(4, 6))
        ttk.Button(t, text="🔷 Ejecutar BFS", command=self._on_bfs).grid(row=2, column=0, sticky="w")

        tk.Label(t, text="DFS — Búsqueda en Profundidad", font=("Segoe UI", 10, "bold")).grid(row=3, column=0, sticky="w", pady=(16, 0))
        self.cb_dfs = ttk.Combobox(t, values=opts, state="readonly", width=18)
        self.cb_dfs.set(PLACEHOLDER)
        self.cb_dfs.grid(row=4, column=0, sticky="w", pady=(4, 6))
        ttk.Button(t, text="🔶 Ejecutar DFS", command=self._on_dfs).grid(row=5, column=0, sticky="w")

    def _on_bfs(self):
        inicio = self.cb_bfs.get()
        if inicio == PLACEHOLDER:
            self._err("Selecciona una ciudad inicial."); return
        visitados, cola, vistos = [], [inicio], {inicio}
        while cola:
            actual = cola.pop(0)
            visitados.append(actual)
            for v in self.grafo.vertices[actual]:
                if v not in vistos:
                    vistos.add(v); cola.append(v)
        self._info(f"BFS desde {inicio} — {len(visitados)} ciudades: " + " → ".join(visitados))
        self._h_origen, self._h_destino = None, None
        self._h_camino, self._h_color_camino = visitados, COLOR_INFO
        self._dibujar()

    def _on_dfs(self):
        inicio = self.cb_dfs.get()
        if inicio == PLACEHOLDER:
            self._err("Selecciona una ciudad inicial."); return
        visitados = []

        def _rec(nodo, vistos):
            vistos.add(nodo); visitados.append(nodo)
            for v in self.grafo.vertices[nodo]:
                if v not in vistos:
                    _rec(v, vistos)

        _rec(inicio, set())
        self._info(f"DFS desde {inicio} — {len(visitados)} ciudades: " + " → ".join(visitados))
        self._h_origen, self._h_destino = None, None
        self._h_camino, self._h_color_camino = visitados, COLOR_AVISO
        self._dibujar()

    def iniciar(self):
        self.root.mainloop()


VentanaEcoVuelo(grafo).iniciar()
```
